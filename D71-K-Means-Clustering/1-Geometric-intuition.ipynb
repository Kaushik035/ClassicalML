# K-Means Clustering Algorithm

- K-Means is an unsupervised machine learning algorithm used for clustering data into distinct groups based on their features. The algorithm aims to partition the dataset into K clusters, where each data point belongs to the cluster with the nearest mean.

- In 2D space, K-Means can be visualized as follows:



- We can easily see the clusters when in 2D space, but in higher dimensions, it becomes more challenging to visualize the clusters. However, the algorithm still works effectively and can be applied to datasets with many features.

# Steps of the K-Means Algorithm

1. **Initialization**: Choose the number of clusters (K) and randomly select K data points as initial centroids.

    

    - In the image, the red, blue, and green points represent the initial centroids (centres) of the clusters.

2. **Assignment of Clusters**:

    - Calculate the distance between each data point and the centroids.
    - Assign each data point to the nearest centroid, forming K clusters.

    

    

3. **Update Centroids**:

    - After assigning all data points to clusters, recalculate the centroids by taking the mean of all data points in each cluster.

    - i.e take the average of all the features of the data points in each cluster to find the new centroid.

    
    (the triangles represent the new centroids after recalculating)

4. **Repeat**:
    - If the centroids have changed from the previous iteration, repeat steps 2 and 3 until the centroids no longer change significantly or a maximum number of iterations is reached.





- So, we just used Euclidean distance to calculate the distance between the data points and the centroids. Which is applicable in all the dimensions. Hence, the K-Means algorithm can be applied to datasets with many features

# How to Choose the Number of Clusters (K)

- We can use the Elbow Method to determine the optimal number of clusters (K). 

- In the Elbow Method, we plot the WCSS (Within-Cluster Sum of Squared/ also known as inertia)  V/S the number of clusters (K).

- The WCSS is the sum of squared distances between each point and its assigned centroid.



- For multiple clusters, we plot the sum of WCSS for each clusters in the y-axis and the number of clusters in the x-axis. 



- The optimal number of clusters is usually at the "elbow" point of the graph, where the WCSS starts to decrease at a slower rate. This indicates that adding more clusters does not significantly improve the clustering.





- This relation always holds true for the WCSS and the number of clusters (K): WCSS decreases as the number of clusters (K) increases. This is because as we increase the number of clusters, the data points are closer to their respective centroids, resulting in a lower WCSS.
