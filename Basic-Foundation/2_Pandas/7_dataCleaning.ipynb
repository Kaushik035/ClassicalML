```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('./data/data_train.csv',nrows=1000)
df.head()
```

# Dorp irrelevant columns

```python
df = df.drop(columns=['contact_type','contact_count'])
df.head()
```

# Handle missing data

1. If at any row any of the data is missing, we can drop that row using dropna() function. This will remove all the rows which have any missing data in any of the columns.

```python
print("Shape of the data before dropping missing values: ", df.shape)
df =df.dropna()
print("Shape of the data after dropping missing values: ", df.shape)
```

2. Drop only the rows which have missing data in specific columns using subset parameter of dropna() function. This will remove all the rows which have any missing data in the specified columns.

```python
df = df.dropna(subset=['age','job'])
```

3. Filling the missing data with some value using fillna() function. This will replace all the missing data in the DataFrame with the specified value.

```python
df = df.fillna(value={'age':df['age'].mean(),'job':'unknown'}) # pass a dictionary to fillna() function to fill different columns with different values
```

4. Use replace() function to convert all the NULL strings  in the whole dataset to NaN values. Then drop using dropna() function.

```python
df = df.replace('NULL', np.nan) # replace NULL strings with NaN values
# or 
# df = df.replace({'NULL': np.nan}) # we can mention multiple values to be replaced in a dictionary format
df = df.dropna()
```

# Fix incosistent values
- Use of replace() function {Very Important}

```python
df['marital_status'] = df['marital_status'].replace({'single':'Single','married':'Married','divorced':'Divorced'}) # replace inconsistent values in a column using replace() function
```

# Standardize text

```python
df['marital_status'] =  df['marital_status'].str.upper() # convert all the values in a column to uppercase using str.upper() function
df.head()
```

# Fix data types

```python
df['age'] = df['age'].astype(int) # convert the data type of a column using astype() function
df.head()
```

# Remove duplicates rows

```python
df = df.drop_duplicates() # drop duplicate rows using drop_duplicates() function
```
