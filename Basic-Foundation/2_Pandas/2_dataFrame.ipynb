# Dataframe

- A dataframe is a two-dimensional, size-mutable, and heterogeneous tabular data structure with labeled axes (rows and columns).
- You can imagine it as an excel spread sheet
- It is commonly used in data analysis and manipulation tasks. In Python, the `pandas` library provides a powerful implementation of dataframes.



````python
import pandas as pd
````

# Dictionary to DataFrame

````python
data = {"Name": ["Alice", "Bob", "Charlie"],
        "Age": [25, 30, 35],
        "City": ["New York", "Los Angeles", "Chicago"]}

df =  pd.DataFrame(data) # D is capitalized because it is a constructor, not a function
````

````python
df
````

````python
# setting custom index for the rows of the dataframe
df1 = pd.DataFrame(data, index = range(1,len(data["Name"]) +1))
df1
````

````python
df2 = pd.DataFrame(data, index = ["Employee1", "Employee2", "Employee3"])
df2
````

````python
print(df2.loc["Employee1"]) # printing the first row using loc
````

````python
print(df2.iloc[0]) # printing the first row using iloc
````

## Add a new column

````python
df2["Job" ] = ["Chef","N/A","SDE"]
df2
````

## Add a new row

````python
new_row_dict = {"Name": ["David", "Eva"],
                         "Age": [28, 32],
                         "City" : ["Houston", "Phoenix"],
                         "Job": ["Engineer", "Designer"]}

new_rows = pd.DataFrame( new_row_dict,index=["Employee4", "Employee5"])

new_rows
````

- Then concatinate the rows 
```
pd.concat([df2, new_rows])
```

````python
df2 = pd.concat([df2, new_rows])
df2
````

````python
df2.iloc[0,1] # prints the value at the 0th row and 1st column
````

````python
df2.iloc[1:4, 0:2] # prints the values from 1st to 3rd row and 0th to 1st column
````

````python
df2.iloc[:,-1] # prints all the rows and only the last column
````

````python
df2.loc[["Employee1", "Employee3"], ["Name", "City"]] # prints the values of Name and City for Employee1 and Employee3
````
