- Filtering based on a single condition

```python
aged_df = df[df['age'] > 50] # filtering the data where age is greater than 50
```

```python
aged_df.head(10)  # displays the first 10 rows of the filtered dataframe
```

```python
management_df = df[df['job'] == 'management'] # filtering the data where job is management
management_df.head(10)  # displays the first 10 rows of the filtered dataframe
```

```python
# people with age greater than 50 and job is management
aged_management_df = df[(df['age'] > 50) & (df['job'] == 'management')]
aged_management_df.head(10)  # displays the first 10 rows of the filtered dataframe
```

```python
# people with age greater than 50 OR job is management
aged_or_management = df[(df["age"] > 50) | (df["job"] == "management")]
aged_or_management.head(10)
```
