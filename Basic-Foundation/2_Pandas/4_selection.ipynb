## Setting default columns and rows to display in pandas

```python
df2 = pd.read_csv('./data/data_train.csv', nrows=100,index_col=0) # using column 0 as index column
df2
```

```python
df2.loc[[40,30], ["marital_status", "job"]]  # selects first 5 rows and columns age and job
```

```python
df2.loc[40:50, ["marital_status", "job"]]  # this would select all the rows with age between 40 and 50 and columns marital_status and job
# But it is giving error because the age (index) have duplicates. So, we cannot use slicing with loc when the index has duplicates.
```

```python
df.loc[0:5, ["age", "job"]]  # selects first 6 rows and columns age and job
# see here 5th row is included, because 5 is the label not index in loc[0:5, ["age", "job"]]
```

```python
df.iloc[0:5, 0:2] # now here 5th row and 2nd col is excluded because iloc is index based selection. So, it is selecting rows from 0 to 4 and columns from 0 to 1.
```
