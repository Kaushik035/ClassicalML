# Aggregate Functions
- Reduces a set of values into a single summary value
- Used to summarize and analyze data
- Often used with groupby() function 

```python
import pandas as pd
```

```python
df = pd.read_csv('./data/data_train.csv',nrows=100)
df
```

## Aggregate functions on whole DataFrame

```python
print("-----------mean-----------")
print(df.mean(numeric_only=True))
print("-----------median-----------")
print(df.median(numeric_only=True))
print("-----------std-----------")
print(df.std(numeric_only=True))
print("-----------var-----------")
print(df.var(numeric_only=True))
print("-----------min-----------")
print(df.min(numeric_only=True))

print("----------count non-null values-----------")
print(df.count())
```

## Aggregate functions on a single column

```python
df['age'].mean()
```

```python
df['marital_status'].value_counts()
```

## Groupby aggregate functions

```python
group = df.groupby('marital_status')
```

```python
group
```

```python
group["age"].mean()
```

```python
group['age'].min()
```

- Grouping based on multiple columns and applying aggregate functions on them

```python
multi_group = df.groupby(['marital_status','education'])
```

```python
multi_group['age'].mean()
```
