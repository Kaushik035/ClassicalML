# Series
- Series is a one-dimensional labeled array capable of holding any data type. It can be created from a list, dictionary, or an array. The axis labels are collectively referred to as the index.

## List to series

```python
#  List to series
data = [101, 201, 301, 604, 456]

series1 = pd.Series(data)

print(series1)
```

```python
data2 = [102.4,12, 13, 14, 15] # it will store all the values as float because one of the value is float
series2 = pd.Series(data2)
print(series2)
```

```python
data3 = ['a', 'b', 'c', 'd', 'e']
series3 = pd.Series(data3)
print(series3)
```

- We can set our own index, but if we don't, it will automatically assign an index starting from 0.
- Use the index parameter to set your own index.
- It accepts a list, array, or series as an argument.

```python
series4 = pd.Series(data, index=['a', 'b', 'c', 'd', 'e'])
print(series4)
```

### Using loc method to access and modify series

- Using the loc method, we can access the elements of a series using their index labels. We can also modify the values of a series using the loc method.

```python
series4.loc['a']
```

```python
print(series4.loc['a'])
```

```python
series4.loc['a'] = 1000 
print(series4.loc['a'])
```

```python
print(series4.loc[['a', 'b', 'c']])
```

### Using iloc method to access and modify series
- Using the iloc method, we can access the elements of a series using their integer index positions. We can also modify the values of a series using the iloc method.

```python
print(series4.iloc[0]) # access the first element of the series using iloc method
```

```python
print(series4.iloc[[0, 1, 2]])
```

```python
print(series4.iloc[:3]) # slice from the 0th index to 2nd index
# index 3 is not included in the slice
```

## Filtering in series

```python
print(series4)
```

```python
print(series4[series4 <= 600]) # filter the series to get the values less than or equal to 600
```

## Dictionary to series

```python
price = {'apple': 52, 'banana': 89, 'orange': 43, 'grapes': 67} # we use the keys as the labels
series5 = pd.Series(price)
```

```python
print(series5)
```

```python
print(series5.iloc[1:3]) # slice the series to get the values from index 1 to 2
```

```python
print(series5.loc['apple']) # access the value of apple using the label
```

```python
print(series5[series5 < 50])
```

```python
arr = [52, 89, 43, 67]
series6 = pd.Series(arr, index = range(1,len(arr)+1)) # we use the range function to create the index
print(series6)
```
