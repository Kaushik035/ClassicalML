# Additive Modelling

- In ML we basically find a function that maps the input features to the target variable.



- Sometimes, the data is complex and we can't find a single function that maps the input features to the target variable.

- For example in the above image, the function in THE RHS, if you try to fit it with a n-degree polynomial,  **Runge's phenomenon** will occur

- **Runge's phenomenon** is a problem of oscillation at the edges of an interval that occurs when using polynomial interpolation with polynomials of high degree.

- So, we can use **Additive Modelling** to solve this type of problem where we can't find a single function that maps the input features to the target variable.

- **Additive Modelling**: Iss technique mei hum complex function ko multiple simple functions mei tod dete hai, aur unko add kar dete hai.

- i.e. A complex function can be represented as a sum of multiple simple functions.

- And boosting also we do this, where this simple function is a base model (Decision Stump or Decision Tree with max_leaves= [8 to 32]).

- That's why on adding more and more base models, we reach towards the actual function that maps the input features to the target variable and error decreases.
