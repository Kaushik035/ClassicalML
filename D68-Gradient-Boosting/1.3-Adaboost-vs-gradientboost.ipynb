# Difference between Gradient Boosting and Adaboost

### 1. Max Leaf Nodes:

- IN Adaboost, the base models are desicion stumps, which means they have only 2 leaf node.

- In Gradient Boosting, the base models are desicion trees with `max_leaves= [8 to 32]` (For samller datasets, we can use `max_leaves= 8` and for larger datasets, we can use `max_leaves= 32`)

### 2. Learning Rate:

- In Adaboost, the learning rate is calculated using the formula:
    $$ \alpha = 0.5 * log(\frac{1 - error}{error}) $$
    where error is the error of the base model.

    - And it is different for each base model so it tells which base model is more important, and it is calculated using the error of the base model.

- In Gradient Boosting, the learning rate is a hyperparameter that is set by the user, and it is same for all the base models.
