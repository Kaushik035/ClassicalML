# Stacking Ensemble Learning

- It is very similar to Voting Ensemble Learning, but the difference is that in Stacking Ensemble Learning, we will use a meta-model to combine the predictions of the base models instead of using majority voting (for classification) or averaging (for regression).



- So, the output from the base models will be used as input features for the meta-model, and the meta-model will learn how to combine the predictions of the base models to make a final prediction.

# Problem with Stacking Ensemble

- If the base models are trained on the entire training dataset and their predictions on the same dataset are used to train the meta-model, the base models may produce overly accurate predictions due to overfitting.

- Consequently, the meta-model learns from these overly optimistic predictions and may fail to generalize well to unseen data.

# Solutions to the Problem

- There are two common solutions to this problem:
    1. **Holdout Method**: Split the training dataset into two parts. Train the base models on the first part and the other part of the training dataset is used to generate predictions from the base models, which are then used to train the meta-model.
        - It is known as **Blending**.

    2. **K-Fold Cross Validation**: Use K-Fold Cross Validation to generate Out-of-Fold (OOF) predictions for training the meta-model.
        - It is known as **Stacking**.

# K-Fold Cross Validation Approach (Stacking)





- Suppose we use **K = 4**, we split the training dataset into 4 folds. For each fold, we train the base models on the other 3 folds and generate predictions for the current (held-out) fold.

- We repeat this process for all 4 folds and all the base models. Finally, we obtain a new dataset in which the predictions from each base model serve as features, and the actual target variable serves as the target variable for training the meta-model.



- The shape of the new dataset will be (N, M+1) = (800,4), where N is the number of rows in the training dataset and M is the number of base models, and 1 is the actual target variable.
