# Difference b/w Bagging and Boosting

### 1. Type of model used

- In **Bagging**, we use model which has **Low Bias and High Variance** (Eg. Fully grown desicion trees, KNN with low k,SVM). So that the model can learn the data well and then we can reduce the variance by using Bagging technique.

- In **Boosting**, we use model which has **High Bias and Low Variance** (Eg. Decision Stump).  So the earlier underfitted model can be improved by using Boosting technique.

### 2. Sequence of model training

- In **Bagging**, all the weak learners are trained in **parallel**. So each weak learner is independent of other weak learners.

- In **Boosting**, all the weak learners are trained in **sequence**. So each weak learner is dependent on other weak learners. The next weak learner will try to correct the mistakes of previous weak learners.

### 3. Weightage of base learners

- In **Bagging**, all the weak learners are given **equal weightage**. So each weak learner will have equal contribution in final prediction.

- In **Boosting**, the weak learners are given **unequal weightage** ($\alpha$ values). The weak learners that perform better are given more weightage, while the weak learners that perform poorly are given less weightage.
