# Core Idea of Voting Ensemble

- The core idea of Voting Ensemble is to combine the predictions of multiple different models to make a final prediction.

- For classification, the final prediction is made by taking a majority vote of the predictions made by the base models.

- For regression, the final prediction is made by taking the mean of the predictions made by the base models.



# Assumptions for Voting Ensemble to work

1. All the base models should be independent of each other. If the base models are correlated, then the voting ensemble will not work as expected.

2. The minimum accuracy of the base models should be 51%, uske neech faida nhi hoga, combined model ka acuuracy will be worst than the worst base model.

# Why Voting works?
- Or we can particullarly say why **Ensemble works**, because this step of acuuracy calculation will be done in all the ensemble methods, so we will see this in bagging, boosting and stacking as well.

#### Like the question is, suppose accuracy of 3 modesls are 0.55, 0.60, 0.75, then how the accouracy of the voting ensemble is more (0.85 say)?

- **Non mathematical explanation:** People from different backgrounds have different knowledge, so they will make different mistakes. So when we combine their predictions, the mistakes will cancel out and the final prediction will be more accurate.

- **Mathematical explanation:** Suppose we have 3 models  all with accuracy 0.7, so the probability of making a wrong prediction is 0.3.

    

    - Now since in voting ensemble, we take the majority vote, so consider the cases where atleast 2 models are make correct preciction (0.7 accuracy)
    - The cases are:
        - All 3 models are correct
        - 2 models are correct, 1 is wrong
    - Now add the probabilities of all the cases where atleast 2 models are correct, we get the final accuracy of the voting ensemble as 0.784, which is more than the accuracy of the base models (0.7).

    - Also you can see if accuracy of the base models is less than 0.5, then the accuracy of the voting ensemble will be less than the accuracy of the base models. So the minimum accuracy of the base models should be 0.51 for voting ensemble to work.

    - Ex. all the models have accuracy 0.4, then the probability of making a wrong prediction is 0.6.
        - So the cases where atleast 2 models are correct are:
            - All 3 models are correct
            - 2 models are correct, 1 is wrong

        - Now add the probabilities of all the cases where atleast 2 models are correct, we get the final accuracy of the voting ensemble as 0.352, which is less than the accuracy of the base models (0.4).
