# Voting classifier





- So, in the above example we used 2 different models, one is Gausian Naive Bayes model and the other is Logistic Regression model. 

- After both the algorithm train independently on the training data, we will collectively pool and make a new voting classifier model 

- You can see that the accuracy of the voting classifier is better than both the base models.



# Hard Voting Classifier (Default)

- In hard voting classifier, the final prediction is made by taking a majority vote of the predictions made by the base models.



# Soft Voting Classifier

- In soft voting classifier, the final prediction is made by taking the mean of the predicted probabilities of the base models.




### What to use Hard Voting or Soft Voting?





- Try both and see which one works better for your dataset. 

- Also sometimes combined model accuracy is less than the best base model, so pick the best base model and use it as your final model.
