# Predicting which class a new data point belongs to using SVM



- We use a vector perpendicular (W in the image) to the hyperplane to determine which side of the hyperplane a new data point lies on.
- If the dot product of the new data point (vector u in the image) on the vector W + b is greater than 0, then the new data point belongs to class 1. Otherwise, it belongs to class -1/ or 0.
- The vector thing is helpful in high dimensional space where we cannot visualize the hyperplane. 

### Desicion Rule
- If W.u + b >= 0, then the new data point belongs to class 1.
- If W.u + b < 0, then the new data point belongs to class -1.
