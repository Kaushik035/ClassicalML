# SVM (Support vector Machine)



- It is clear that we would choose $pi_1$ over $pi_2$, because it will work better on unseen data

- Support Vector Machine (SVM) is a supervised machine learning algorithm used for classification and regression tasks.
- It tries to find the best boundary known as hyperplane that separates different classes in the data. It is useful when you want to do binary classification like spam vs. not spam or cat vs. dog.

- The main goal of SVM is to maximize the margin between the two classes. The larger the margin the better the model performs on new and unseen data.


# Core Idea

- We are trying to find the **Margin maximizing hyperplane**
- i.e. we are trying to find the $W$ and $b$, that maximized $d$ (margin) in the below picture



## Steps:
1. Choose a hyper plane $pi$
2. Then move to both direction (+ve and -ve) till you get the first touching point (known as support vectors) and draw parallerl planes to $pi$ ($pi_+$ and $pi_-$)
3. Now our objective is to find a plane that maximizes the margin $d$



# Concept of Support Vectors

- Support vectors are the data points that are closest to the hyperplane and influence its position and orientation. These points are critical in defining the optimal hyperplane.



# Benifit of using SVM

1. It is robut to outliers because it only considers the support vectors and ignores the rest of the data points.
2. Works well for non-linear data using kernel trick.
3. Can be used for both classification and regression tasks.
