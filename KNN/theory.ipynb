# Steps to implement KNN Algorithm:

1. Choose the number of neighbors (K) you want to consider for making predictions.

2. For the point you want to classify, calculate the distance between that point and all other points in the dataset. Common distance metrics include Euclidean distance, Manhattan distance, etc.

3. Sort the distances in ascending order and select the K nearest neighbors.

4. Determine the majority class among the K nearest neighbors. 

- So, the class with the highest frequency among the neighbors will be assigned to the point you want to classify.

# How to select the value of K?

- There are 2 ways to determine value of k:
    1. Heuristic approach (padha likha jugaad): 
        - k = root of n, where n is the number of data points in the training dataset.
        - But refrain from taking even values of k, as it may lead to ties in the voting process.
    
    2. Experimentation approach (preferred):
        - Cross- validation:
            

            - Here we choose different values of k, and run the KNN model, and the value of k for which we get the best results (accuracy.. or other metrices) we choose that value of k

# Desicion Surface/ boundary

- It is a tool used to understand the performance of a **classification algorithm**



- It is basically representation of the desicion boundary created after the classification algo

- Mostly used for 1D, 2D, 3D data as easy to visualize

## How to generate it? (No need to look in detail)

- Then send the points in the mesh grids -> KNN

- Better is use `mlextend library` to generate the desicion surface



# Overfitting and Underfitting in KNN



- **Smaller the value of k**: It will capture even the minor changes in its nearby neighbour, so it will create small pools of desicion regions based on the nearby values

- **Larger the value of k**: (Max value of k = no. of data points), So it will always predict the majority class present in the whole dataset







# Limitations of KNN

1. Not preferred on Large datasets

    - KNN is a **lazy learning technique**
    - It means most of the work happens at inference/prediction time
    - So before getting a query point, we just store the training points 
    - And as we get the query point, then we calculate distance and do further steps of KNN
    - So latency will be high

2. High dimensional data
    - Because of **curse of dimensionality*, distance calculations are not reliable
    - So if distances are distorted in KNN, then results will be distorted

3. Doesn't work well with outliers





- The outlier region should be blue, but for k =3, it became pink

4. Non Homogenous feature scales



5. Imbalanced Datasets



6. KNN is not good for inference, it is good for prediction only
    - Like we can predict the class using KNN
    - But konse feature the kitna weightage h that we cant get from KNN
    - So basically KNN behaves like a black box
