# Hyperparameters of Decision Trees

- A very important hyperparameter is the **depth of the tree**. This controls how deep the tree can grow, which can help prevent overfitting.

# Overfitting in Decision Trees

- Performs well on training data but poorly on unseen data.



- Initially, we have 200 data points at the root node.
- The decision tree splits the data based on certain features:
  - First split: 200 → 140 + 60
  - Second split: 140 → 100 + 40
  - Further split: 100 → 98 + 2 (illustrative interpretation of the diagram).
- The tree keeps making decisions and splitting nodes into smaller subsets.
- If we continue splitting until every leaf node contains only one or very few data points, the tree may start learning noise, outliers, and erroneous data instead of actual patterns.
- This leads to overfitting:
  - Performs very well on training data.
  - Performs poorly on unseen/test data.
- When `max_depth=None`, there is no predefined maximum depth, so the tree can keep growing until another stopping condition is met.
- The tree may become unnecessarily complex and memorize the training dataset.

# Underfitting in Decision Trees



- So for the new query point the model predicted the class with the majority of the points in the region, but we didnt split the data enough to capture the underlying patterns in the data, which means that the model is underfitting to the training data and not capturing the complexity of the data.



- We hypertune the **max_depth** parameter to find the optimal depth of the tree that balances bias and variance.

# How to prevent overfitting?

1. Set a maximum tree depth (`max_depth`).
2. Set minimum samples required to split (`min_samples_split`).
  - It controls the minimum number of samples required to split an internal node. If a node has fewer samples than this threshold, it will not be split further.
  - Lower the `min_samples_split`, the more complex the tree can become, which may lead to overfitting. Conversely, increasing this value can lead to underfitting if set too high, as it may prevent the tree from capturing important patterns in the data.
3. Set minimum samples required at a leaf (`min_samples_leaf`).
  - It controls the minimum number of samples required at a leaf node. If a leaf node has fewer samples than this threshold, it will not be created.

4. Use pruning to remove unnecessary branches.

5. When working with high-dimensional data, consider using feature selection techniques to reduce the number of features before training the decision tree. (`max_features` parameter can be used to limit the number of features considered for splitting at each node).
  - Suppose you have 2 features, and you set `max_features=1`. This means that at each split, the decision tree will only consider one of the two features for splitting.
  - where this feature is selected randomly from the available features. This introduces randomness into the tree-building process, which can help reduce overfitting by preventing the model from relying too heavily on any single feature.
  - However if you set `max_features=2`, the decision tree will consider both features for splitting at each node, and will calulate Information Gain for both features and select the one that provides the best split. This can lead to a more complex tree that may overfit the training data, especially if the dataset is small or noisy.

6. `max_leaf_nodes` parameter can be used to limit the number of leaf nodes in the tree. By setting a maximum number of leaf nodes, you can control the complexity of the tree and prevent it from growing too large and overfitting the training data.

7. `Min_impurity_decrease` parameter can be used to set a threshold for the minimum impurity decrease required to split a node. If the impurity decrease from a potential split is less than this threshold, the split will not be made. This can help prevent the tree from making splits that do not significantly improve the model's performance, thus reducing overfitting.
