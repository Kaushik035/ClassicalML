# Entropy v/s Probability



- The entropy is maximum when the probability of all classes is equal. For example, if we have 2 classes and the probability of each class is 0.5, then the entropy will be maximum.

- For the extreme case, if the probability of one class is 1 and the other class is 0, then the entropy will be minimum (0). This means that there is no uncertainty in predicting the output.
