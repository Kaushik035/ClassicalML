# Bagging Tips

- Bagging generally gives better results than Pasting because in bagging, we sample with replacement, so some rows will be sampled multiple times and some rows will not be sampled at all. This will create more diversity among the base models and will reduce variance of the combined model.

- Good results come around the 25% to 50% of the rows sampled for each base model. (`max_samples` parameter [0.25, 0.5])

- Random patches and random subspaces should be used while dealing with high dimensional data.

- To find the correct hyperparameters, we can use **grid search CV** or **random search CV** (which is like loop chalake saare hyperparameters ko try krna). We can also use cross-validation to find the best hyperparameters.
