# Ensemble Learning

- The word "Ensemble" refers to group of something.

- In machine learning, ensemble learning is a technique that combines multiple models to improve the overall performance and accuracy of predictions.

- The fundamental of Ensemble Learning is based on the fact of "Wisdom of the Crowd".

# Wisdom of the Crowd

- The concept of "Wisdom of the Crowd" suggests that the collective opinion of a group of individuals can be more accurate than that of a single expert.


# Core Idea

- The core idea of ensemble learning is to leverage the strengths of multiple models to make better predictions than any individual model could achieve on its own.

- There are two phases in any machine learning model: training and prediction.

- Ensemble learning can be applied in both phases to enhance the performance of the model.

- Let's now see for the prediction phase how ensemble learning can be applied. Later, we will discuss how it can be applied in the training phase.

# Types of Ensemble Learning

- There are 4 main types of ensemble learning techniques:

1. Voting Ensemble
2. Bagging Ensemble
    - Random Forest is a type of Bagging Ensemble
3. Boosting Ensemble
    - Adaboost, Gradient Boosting, XGBoost, LightGBM, CatBoost are types of Boosting Ensemble
4. Stacking Ensemble

### What is the basis of these types?

- As we discussed earlier, the base models should be different from each other. The difference can be in terms of algorithms or the data used for training.

- So these 4 types of ensemble learning techniques are based on how the base models are made different from each other.

### 1. Voting Ensemble

- In Voting Ensemble, the base models are made different from each other by using different algorithms. They all are trained on the same training data.

- While predecting, the final prediction is made by taking the majority vote of all the base models or in case of regression, the final prediction is made by taking the average of all the predictions made by the base models.



### 2. Stacking Ensemble

- In Stacking Ensemble, the base models are made different from each other by using different algorithms. They all are trained on the same training data.

- But the difference between Stacking and Voting Ensemble is that in Stacking Ensemble, the predictions made by the base models are used as input features for a new model called the meta-model or blender model.

- The meta-model gets trained on the predictions made by the base models and assigns weights to each base model based on their performance. The final prediction is made by the meta-model.

- Where as in voting, all the base models are given equal weightage and the final prediction is made by taking the majority vote or average of all the predictions made by the base models.



### 3. Bagging Ensemble

- Bagging stands for Bootstrap Aggregating.
- In Bagging Ensemble, the base models are made different from each other by using the same algorithm but training them on different subsets of the training data.

- Bootstraping is a statistical technique that involves sampling with replacement from the original dataset to create multiple subsets of the data. And it is highly likely that the all the subsets will be different from each other as we are doing ranqdom sampling with replacement.



- Prediction is made by taking the majority vote or average of all the predictions made by the base models.

- A special case of Bagging Ensemble is Random Forest, where the base models are decision trees.

### 4. Boosting Ensemble

- It is the most powerful and useful ensemble learning technique.

- In Boosting Ensemble, the base models are made different from each other by using the same algorithm but training them sequentially on the same training data.

- The first base model is trained on the original training data and the subsequent base models are trained on the same training data but with a focus on the misclassified samples from the previous base model.

- i.e Previous model next model ko bata deta hai ki usne kaha kaha galti ki hai aur next model us galti ko sudharne ki koshish karta hai.

# Disadvantages of Ensemble Learning

- Increased complexity in terms of model selection, training, and deployment.

# Advantages of Ensemble Learning

1. Improvement in performance

2. Bias and variance reduction (Low Bias and Low Variance is the best case scenario in machine learning)

    - Ensemble learning helps in helps in achieving the hard task of Bias Variance tradeoff.

    - So for a Low Bias and High Variance model, we can use Bagging Ensemble to reduce the variance maintaining the low bias, and for a High Bias and Low Variance model, we can use Boosting Ensemble to reduce the bias maintaining the low variance.

3. Robustness 
    - Model alag alag data k liye robust ho jata hai. So even changing the data distribution will not affect the performance of the model much.
