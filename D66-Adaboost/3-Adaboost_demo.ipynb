```python
import pandas as pd
import numpy as np
from mlxtend.plotting import plot_decision_regions
```

```python
df = pd.DataFrame()
```

```python
df['X1'] = [1,2,3,4,5,6,6,7,9,9]
df['X2'] = [5,3,6,8,1,9,5,8,9,2]
df['label'] = [1,1,0,1,0,1,0,1,0,0]
```

```python
df
```

```python
import seaborn as sns
sns.scatterplot(x=df['X1'],y=df['X2'],hue=df['label'])
```

- Step 1: Assign equal weights to all the training samples. The weight of each sample is 1/N, where N is the number of training samples.

```python
df['weights'] = 1/df.shape[0]
```

```python
df
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt1 = DecisionTreeClassifier(max_depth=1) # decision stump
```

```python
X = df.iloc[:,0:2].values
y = df.iloc[:,2].values
```

```python
# Step 2 - Train 1st model
dt1.fit(X,y)
```

```python
from sklearn.tree import plot_tree
plot_tree(dt1)
```

```python
plot_decision_regions(X, y, clf=dt1, legend=2)
```

```python
df['y_pred'] = dt1.predict(X)
```

```python
df
```

- Alpha calculation

```python
def calculate_model_weight(error):

  return 0.5*np.log((1-error)/(error))
```

```python
# Step 3 - calculate model weight
alpha1 = calculate_model_weight(0.3)
alpha1
```

- Update the weights of the training samples based on the predictions of the weak learner. The weights of the misclassified samples are increased, while the weights of the correctly classified samples are decreased.

```python
# Step 4 - Update weights
def update_row_weights(row,alpha=0.423):
  if row['label'] == row['y_pred']:
    return row['weights'] * np.exp(-alpha)
  else:
    return row['weights'] * np.exp(alpha)
```

```python
df['updated_weights'] = df.apply(update_row_weights,axis=1)
```

```python
df
```

```python
df['updated_weights'].sum()
```

- Normalize the weights of the training samples so that they sum to 1.

```python
df['nomalized_weights'] = df['updated_weights']/df['updated_weights'].sum()
```

```python
df
```

```python
df['nomalized_weights'].sum()
```

- Make range for each sample based on the weights. The range for each sample is proportional to its weight. More the weight of a sample, more the range for that sample.

```python
df['cumsum_upper'] = np.cumsum(df['nomalized_weights'])
```

```python
df['cumsum_lower'] = df['cumsum_upper'] - df['nomalized_weights']
```

```python
df[['X1','X2','label','weights','y_pred','nomalized_weights','cumsum_lower','cumsum_upper']]
```

- <strong style="color:orange">Now create a new dataset, by generating random numbers between 0 and 1, and selecting the sample corresponding to the range in which the random number falls. This will create a new dataset with the same number of samples as the original dataset, but with some samples being repeated and some being left out.</strong>

### This step is called UPSAMPLING

```python
def create_new_dataset(df):

  indices = []

  for i in range(df.shape[0]):
    a = np.random.random()
    for index,row in df.iterrows():
      if row['cumsum_upper'] > a and a > row['cumsum_lower']:
        indices.append(index)
  return indices
```

```python
index_values = create_new_dataset(df)

index_values
```

- So now the below new dataset will be used to train the next weak learner. And the process will be repeated for all the weak learners. And finally, all the weak learners will be combined to make the final strong learner.

```python
second_df = df.iloc[index_values,[0,1,2,3]] # taking only X1,X2,label,weights columns
```

- We always start with equal weights for all the samples at the beginning of the training process of any weak learner model

```python
second_df
```

<hr>

From here on its recursive for all the weak learners. The below steps will be repeated for all the weak learners.
