# Adaboost Algorithm

- It is a boosting algorithm, quite old algorithm and was used for phase detection in the early 2000s. 

- Adaboost is a good algorithm, but ultimately Random Forest, XGBoost and LightGBM are better algorithms than Adaboost.

- So you we will rarely be applying adaboost in real life, but if you want to learn **Boosting**, then  we have to start with adaboost

- We have to learn about 3 topics before we can understand adaboost algorithm. They are:
    1. Weak Learner
    2. Decision Stump
    3. +1 and -1 labels

### 1. Weak Learner

- A weak learner is a model that is slightly better than random guessing. For example, if we have a binary classification problem, then a weak learner is a model that has an accuracy slightly better than 50%.

- It is related here because adaboost is a boosting algorithm that combines multiple weak learners to create a strong learner (a model that has a high accuracy).

### 2. Decision Stump

- A decision stump is a decision tree with only one level (`depth=1`). It is a weak learner because it can only make a decision based on one feature.

- We choose the split point of the decision stump based on the feature that gives the best split (the feature that gives the highest Information Gain).

- In Adaboost, though we can use any weak learner, we will be using decision stumps as weak learners 99% of the time
    - As it gives the best result
    - Using desicion tree, our model becomes more interpretable and we can visualize the decision stump easily.

### 3. +1 and -1 labels

- In Adaboost, we will be using +1 and -1 labels instead of 1 and 0 labels. This is because the algorithm is mathematically easier to work with when we use +1 and -1 labels.



# The big idea of Adaboost

- Adaboost is a **stage wise additive metheod** (we add the weak learners one by one in a sequential manner)

- Suppose given cgpa and iq, we want to predict whether a student will get placed or not. (+1 means placed and -1 means not placed)



- So, in the above image you can see we are using 3 weak learners (decision stumps) to make the final prediction.

### STEPS:

1. We will first train a weak learner (decision stump) on the training data and the splitting point of the decision stump will be chosen based on the feature that gives the best split (the feature that gives the highest Information Gain i.e entropy decreases the most).

    
 
    - But as you can see, 3 +ve points are misclassified so before we move to the next weak learner, we will **upsample** the misclassified points and give them more weightage so that the next weak learner can focus on them and classify them correctly.

    - Also we **calucualte the weight of the weak learner** ($\alpha_1$) based on its accuracy. The more accurate the weak learner, the more weight it will have in the final prediction.

2. Now go to the next weak learner with the upsampled data and repeat the process. The second weak learner will focus more on the misclassified points from the first weak learner and try to classify them correctly.

    - It have 2 -ve points misclassified, so we will upsample them and give them more weightage so that the next weak learner can focus on them and classify them correctly.

    - Calcualte $\alpha_2$ based on the accuracy of the second weak learner.

3. Now go to the third weak learner with the upsampled data and repeat the process. The third weak learner will focus more on the misclassified points from the second weak learner and try to classify them correctly.

    - As it is the last weak learner, we don't need to upsample the misclassified points. Just calculate $\alpha_3$ based on the accuracy of the third weak learner.



- Finally , we will combine the predictions of all the weak learners using their weights ($\alpha_1$, $\alpha_2$, $\alpha_3$) to make the final prediction.



#### $$Final Prediction = sign(\alpha_1 * h_1(x) + \alpha_2 * h_2(x) + \alpha_3 * h_3(x))$$

where $h_1(x)$, $h_2(x)$ and $h_3(x)$ are the predictions of the weak learners and $\alpha_1$, $\alpha_2$ and $\alpha_3$ are their weights.





- You can see that as model 2 have the highest weightage, so despit of other 2 model s predicting -1, the final prediction is +1 because model 2 have the highest weightage and it predicted +1. (Use the final prediction formula to verify this.)

- Also geometrically, the final desicion boundary is the weighted average of the decision boundaries of the weak learners. The more weightage a weak learner has, the more influence it has on the final decision boundary.
